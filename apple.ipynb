```python
import pandas as pd
import numpy as np
import plotly.express as px
```

```python
df['Date']=pd.to_datetime(df['Date'])
df["year"]=df["Date"].dt.year
df.head()
df["month"]=df["Date"].dt.month_name()
df.head()
```

```python
df["Sale_Event"] = df["Sale_Event"].fillna("No Sale")
```

```python
data = df.groupby(['year','Product_Category'])['Current_Price_USD'].mean().reset_index()
fig=px.line(data,
    x='year',
    y='Current_Price_USD',
    color='Product_Category',
    # markers=True,
    title='Yearly Average current price by Product category'

)
fig.update_layout(
    xaxis_title='month',
    yaxis_title='Average current price(USD)',
    template='plotly_white'
)
fig.show()
```

```python
area_data = (
    df.groupby([df["Date"].dt.to_period("M"), "Condition"])
      .size()
      .unstack(fill_value=0)
)

# Convert PeriodIndex back to Timestamp
area_data.index = area_data.index.to_timestamp()

# Reset index for Plotly
area_data = area_data.reset_index()

# Convert wide format to long format
area_long = area_data.melt(
    id_vars="Date",
    var_name="Condition",
    value_name="Count"
)

# Create Stacked Area Chart
fig = px.area(
    area_long,
    x="Date",
    y="Count",
    color="Condition",
    title="Historical Lifecycle Volume of Product Conditions",
    labels={
        "Date": "Date",
        "Count": "Number of Products",
        "Condition": "Product Condition"
    }
)

fig.update_layout(
    xaxis_title="Date",
    yaxis_title="Product Count",
    template="plotly_white"
)

fig.show()
```

```python
model_df = (
    df.query("Model_Name == 'iPhone 13 Pro (128GB)'")
      .sort_values("Date")
)

# Create Step-Line Chart
fig = px.line(
    model_df,
    x="Date",
    y="Current_Price_INR",
    title="Price Decay Lifecycle of iPhone 13 Pro (128GB)",
    markers=True
)

# Convert line to step-line
fig.update_traces(
    line_shape="hv",   # Step line (horizontal-vertical)
    line=dict(width=3)
)

# Customize layout
fig.update_layout(
    xaxis_title="Date",
    yaxis_title="Current Price (INR)",
    template="plotly_white",
    hovermode="x unified"
)

fig.show()
```

```python
stacked_data = (
    df.groupby(['Product_Category', 'Stock_Status'])
      .size()
      .unstack(fill_value=0)
      .reset_index()
)

# Convert to long format for Plotly
stacked_data = stacked_data.melt(
    id_vars='Product_Category',
    var_name='Stock_Status',
    value_name='Count'
)

# Create Stacked Column Chart
fig = px.bar(
    stacked_data,
    x='Product_Category',
    y='Count',
    color='Stock_Status',
    title='Inventory Liquidation Risk Profile',
    labels={
        'Product_Category': 'Product Category',
        'Count': 'Number of Products',
        'Stock_Status': 'Stock Status'
    }
)

# Stack the bars
fig.update_layout(
    barmode='stack',
    xaxis_title='Product Category',
    yaxis_title='Number of Products',
    legend_title='Stock Status',
    template='plotly_white'
)

fig.show()
```

```python
grouped_data = (
    df.groupby(['Platform', 'Condition'])
      .size()
      .reset_index(name='Count')
)

# Create Side-by-Side Bar Chart
fig = px.bar(
    grouped_data,
    x='Platform',
    y='Count',
    color='Condition',
    barmode='group',
    title='Platform Representation of Specific Hardware Tiers',
    labels={
        'Platform': 'Platform',
        'Count': 'Number of Products',
        'Condition': 'Product Condition'
    },
    text='Count'
)

# Customize the layout
fig.update_layout(
    xaxis_title='Platform',
    yaxis_title='Number of Products',
    legend_title='Condition',
    template='plotly_white'
)

# Display count labels on bars
fig.update_traces(textposition='outside')

fig.show()
```

```python
df["Price_Tier"] = pd.cut(
    df["Launch_Price_USD"],
    bins=3,
    labels=["Budget", "Mid-Range", "Premium"]
)

# Count Stock Status in each Price Tier
stock = (
    df.groupby(["Price_Tier", "Stock_Status"])
      .size()
      .reset_index(name="Count")
)

# Convert counts to percentages
stock["Percentage"] = (
    stock.groupby("Price_Tier")["Count"]
         .transform(lambda x: x / x.sum() * 100)
)

# Plot 100% Stacked Bar Chart
fig = px.bar(
    stock,
    x="Price_Tier",
    y="Percentage",
    color="Stock_Status",
    text=stock["Percentage"].round(1).astype(str) + "%",
    title="Stock Health Index Across Price Tiers",
    labels={
        "Price_Tier": "Price Tier",
        "Percentage": "Percentage (%)",
        "Stock_Status": "Stock Status"
    }
)

fig.update_layout(
    barmode="stack",
    template="plotly_white",
    yaxis_title="Percentage (%)",
    xaxis_title="Price Tier"
)

fig.show()
```
